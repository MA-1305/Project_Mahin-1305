## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Create model output for Week 7 action playbook

import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

# Load the same dataset used in W05
iris = load_iris()

X = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

y = iris.target

# Same split as W05
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# Same model as W05
model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train model
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Prediction confidence
y_probability = model.predict_proba(X_test).max(axis=1)

# Create ranked action queue
actions = X_test.copy()

actions["actual_class"] = y_test
actions["predicted_class"] = y_pred
actions["model_probability"] = y_probability

actions = actions.sort_values(
    "model_probability",
    ascending=False
).reset_index(drop=True)

actions["Rank"] = range(
    1,
    len(actions) + 1
)

# Action
def assign_action(probability):
    if probability >= 0.65:
        return "Review Immediately"
    elif probability >= 0.50:
        return "Review"
    else:
        return "Monitor Performance"

# Reason
def assign_reason(probability):
    if probability >= 0.65:
        return "HIGH_MODEL_SCORE"
    elif probability >= 0.50:
        return "MEDIUM_MODEL_SCORE"
    else:
        return "LOW_MODEL_SCORE"

actions["Action"] = actions["model_probability"].apply(
    assign_action
)

actions["Reason Code"] = actions["model_probability"].apply(
    assign_reason
)

print("Ranked action queue created successfully.")
print()
print(actions.head(20))
```

## Ranked Actions and Reason Codes

The model output from Week 5 is converted into a ranked action queue. Pages with higher model probability are given higher review priority.

The reason code explains why the page received its action category. The queue is intended to support human review rather than automatically determine the final content decision.
